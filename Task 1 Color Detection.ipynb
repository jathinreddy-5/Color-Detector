```python
import cv2
import pandas as pd

# File paths
IMAGE_PATH = "/Users/jathinreddy/Desktop/slash mark/pic1.jpg"
CSV_PATH = "/Users/jathinreddy/Desktop/slash mark/colors.csv"

# Load color dataset
columns = ["color", "color_name", "hex", "R", "G", "B"]

try:
    df = pd.read_csv(CSV_PATH, names=columns, header=None)
except Exception as e:
    print("Error loading CSV:", e)
    raise SystemExit

# Load image
img = cv2.imread(IMAGE_PATH)

if img is None:
    print("Image not found!")
    raise SystemExit

img = cv2.resize(img, (800, 600))

# Global variables
clicked = False
r = g = b = 0

# Find closest color
def get_color_name(R, G, B):
    minimum = float("inf")
    color_name = "Unknown"

    for i in range(len(df)):
        distance = (
            abs(R - int(df.loc[i, "R"])) +
            abs(G - int(df.loc[i, "G"])) +
            abs(B - int(df.loc[i, "B"]))
        )

        if distance < minimum:
            minimum = distance
            color_name = df.loc[i, "color_name"]

    return color_name

# Mouse click callback
def mouse_click(event, x, y, flags, param):
    global clicked, r, g, b

    if event == cv2.EVENT_LBUTTONDOWN:
        clicked = True

        b, g, r = img[y, x]

        b = int(b)
        g = int(g)
        r = int(r)

        print(f"RGB = ({r}, {g}, {b})")

# Create window
WINDOW_NAME = "Color Detector"

cv2.namedWindow(WINDOW_NAME)
cv2.setMouseCallback(WINDOW_NAME, mouse_click)

# Main loop
while True:

    display = img.copy()

    if clicked:
        color_name = get_color_name(r, g, b)

        text = f"{color_name}  R={r} G={g} B={b}"

        cv2.rectangle(
            display,
            (20, 20),
            (780, 70),
            (b, g, r),
            -1
        )

        text_color = (0, 0, 0) if (r + g + b) > 500 else (255, 255, 255)

        cv2.putText(
            display,
            text,
            (30, 55),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            text_color,
            2,
            cv2.LINE_AA
        )

    cv2.imshow(WINDOW_NAME, display)

    key = cv2.waitKey(10) & 0xFF

    if key == 27:  # ESC key
        break

    if key == ord("q"):  # Q key
        break

cv2.destroyAllWindows()

print("Program Closed Successfully")
```
